# 01-Análise exploratória de retornos
 Análise de retornos dos ativos em busca de fenônemos como fat tails e clustering de volatilidade. Busca-se justificar o uso de GARCH e analisar o comportamento da normal aplicada no contexto.

## 1.Carregar os dados
Carregar os dados já filtrados do src/data.py

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

#Leitura dos arquivos parquet criados no data.py 
precos = pd.read_parquet("../data/precos.parquet")
retornos = pd.read_parquet("../data/retornos.parquet")

retornos.head()
```

## 2. Preços e retornos ao longo do tempo

```python
precos.plot(subplots=True, figsize=(12,8)) #Cria vários gráficos com cada ativo do DataFrame
```

Observa-se nos gráficos alguns períodos de clustering/agrupamento de volatilidade, principalmente em épocas de insegurança política.
- Pandemia do Covid (2020): clustering mais claro da série, após o anúncio doo avanço do Covid-19, a bolsa brasileira sofreu um grande impacto, com grandes quedas seguidas de grandes correções.
- Eleições de 2026: fenômeno de agrupamento de volatilidade mais recente, praticamente no fim da série. Ocorreu devido a uma alta expectativa do mercado da eleição do candidato Flávio Bolsonaro, puxando diversos dias de alta, porém após a reeleição de Lula, o mercado sofreu uma forte queda.
- Crise econômica do governo Dilma (2015): ocorreu devido à alta instabilidade política e econômica no país: iminência do impechament da presidente, escândalos de corrupção, pedaladas fiscais, entre outros. Dessa forma, houve uma grande queda seguida de correções.

Também é possível observar alguns impactos mais específicos de ativos:
- Greve dos caminhoneiros (2018): provocou uma grande queda na Petrobras, também seguido de correções com diversas oscilações
- Rompimento da barragem de Brumadinho (2019): provocou uma grande queda na Vale, também agrupando a volatilidade.


```python
for ativo in retornos.columns: #Percorre por todos os ativos do DataFrame
    retornos_ativo = retornos[ativo]
    max_ativo = retornos_ativo.max() # Maior retorno dos ativos
    min_ativo = retornos_ativo.min() #Menor retorno dos ativos
    dp_ativo = retornos_ativo.std() #Desvio padrão dos ativos
    media_ativo = retornos_ativo.mean() #Média do ativo

    pontos_ativo = np.linspace(min_ativo, max_ativo, 500) # 500 nuémros espaçados do inicio ao fim

    normal_ativo = norm.pdf(pontos_ativo, media_ativo, dp_ativo) # Altura normal dos ativos

    plt.figure() # Nova figura

    plt.title(f"Retornos x Normal ({ativo})") # Título
    plt.xlabel("Log-retorno diário") #Legenda do eixo x
    plt.ylabel("Densidade") #Legenda do eixo y
    plt.yscale("log") #Escala do eixoy: vai multiplicando/dividindo por 10, assim podemos enxergar com clarezas os extremos
    plt.ylim(1e-2, 50) #Limites do eixo y para melhor visualização



    plt.hist(retornos_ativo, bins = 100, density=True, alpha=0.6, label="Distribuição dos retornos") # Distribuição dos retornos
    plt.plot(pontos_ativo, normal_ativo, label="Distribuição Normal") #Distribuição normal
    plt.legend() # Para mostrar a legenda

    plt.show() # Para mostrar o gráfico









```

# Análise das ditribuições

A partir de uma interpretação dos gráficos dados, podemos tirar algumas conclusões: 

- A Ibovespa apresenta uma frequencia de dias extremos muito maior quando comparada com os outros dois ativos. Além de apresentar dias com os retornos de maior oscilação.
- Nos 3 ativos apresentam muito mais dias extremos com log-retornos negativos do que log-retornos positivos. Logo, a queda é muito brusca, mas as correções são feitas em altas menos acentuadas.
- A a altura da normal é menor que a distribuição dos retornos, assim como as caudas tem um kurtosis mais elevada. Portanto, temos um comportamento característico de Fat Tails/ Caudas Pesadas.
